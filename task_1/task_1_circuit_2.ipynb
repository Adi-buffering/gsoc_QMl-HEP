```python
import pennylane as qml
import numpy as np

# 5 wires: wire 0 = ancilla, wires 1,2 = state |psi>, wires 3,4 = state |phi>
dev = qml.device('default.qubit', wires=5)

@qml.qnode(dev)
def swap_test_circuit():
    # === State Preparation ===
    # First 2-qubit state |psi> on wires 1, 2
    qml.Hadamard(wires=1)        # H on qubit 0 (first register)
    qml.RX(np.pi / 3, wires=2)   # RX(pi/3) on qubit 1 (second register)

    # Second 2-qubit state |phi> on wires 3, 4
    qml.Hadamard(wires=3)        # H on qubit 2 (third register)
    qml.Hadamard(wires=4)        # H on qubit 3 (fourth register)

    # === SWAP Test ===
    # Step 1: Hadamard on ancilla
    qml.Hadamard(wires=0)

    # Step 2: Controlled-SWAP between corresponding qubits
    qml.CSWAP(wires=[0, 1, 3])   # C-SWAP: ancilla controls swap of wire 1 <-> wire 3
    qml.CSWAP(wires=[0, 2, 4])   # C-SWAP: ancilla controls swap of wire 2 <-> wire 4

    # Step 3: Hadamard on ancilla
    qml.Hadamard(wires=0)

    # Measure probabilities on ancilla qubit
    return qml.probs(wires=0)

# Print the circuit diagram
print("=" * 60)
print("Circuit 2: SWAP Test (PennyLane)")
print("=" * 60)
print(qml.draw(swap_test_circuit)())

# Execute and get results
probs = swap_test_circuit()
p0 = probs[0]  # Probability of measuring |0> on ancilla
p1 = probs[1]  # Probability of measuring |1> on ancilla


print(f"\nMeasurement Results:")
print(f"  P(|0⟩) = {p0:.6f}")
print(f"  P(|1⟩) = {p1:.6f}")
```
