```python
import cirq
import numpy as np

# Create 5 qubits
qubits = cirq.LineQubit.range(5)
circuit = cirq.Circuit()

# Apply Hadamard on every qubit
circuit.append(cirq.H.on_each(*qubits))

# Apply CNOT on pairs: (0,1), (1,2), (2,3), (3,4)
for i in range(4):
    circuit.append(cirq.CNOT(qubits[i], qubits[i + 1]))

# SWAP qubits 0 and 4
circuit.append(cirq.SWAP(qubits[0], qubits[4]))

# Apply RX(pi/2) on qubit 2
circuit.append(cirq.rx(np.pi / 2)(qubits[2]))

# Print text diagram
print("=" * 60)
print("Circuit 1: 5-Qubit Cirq Circuit")
print("=" * 60)
print(circuit)

# SVG visualization (works in Jupyter/Colab)
from cirq.contrib.svg import SVGCircuit
SVGCircuit(circuit)
```
